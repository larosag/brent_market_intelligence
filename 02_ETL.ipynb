```python
import yfinance as yf
import pandas as pd
```

```python
print(yf.__version__) 
```

```python
ticker = 'BZ=F'
periodo = '1y'
intervalo = '1d'

datos_brutos = yf.download(
    ticker,
    period=periodo,
    interval=intervalo
)

datos_brutos = datos_brutos.reset_index()
datos_brutos.columns = [c[0].lower() for c in datos_brutos.columns]

# renombro las columnas

datos_brutos.rename (columns={
    'date':'fecha', 
    'open':'apertura', 
    'high':'maximo', 
    'low':'minimo', 
    'close':'cierre', 
    'volume':'volumen'}, inplace=True
    )



datos_brutos.head()
```

```python
datos_brutos.info()
```

```python
# Ingestión de datos
from pyspark.sql import SparkSession

# Cluster (iniciar Spark)

spark = SparkSession.builder.getOrCreate()

spark 
```

```python
# Convertir un DF Pandas ---> Spark

df_spark = spark.createDataFrame(datos_brutos)

# Mostrar el DF Spark

display(df_spark)

(
    df_spark.write
    .format('delta')
    .mode('overwrite')
    .option('overwriteSchema', 'true')
    .saveAsTable('db_analytics.bronze')
)

# print(f'Total registrados: {df_spark.count()}')
```
