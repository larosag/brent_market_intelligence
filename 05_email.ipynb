```python
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI
import os


load_dotenv()
```

```python
# leer prompt

with open("prompts/html.txt", "r") as archivo:
    prompt_template = archivo.read()
```

```python

analisis = dbutils.jobs.taskValues.get(
    taskKey = 'agente_analisis',
    key = 'analisis'
)
```

```python
prompt = prompt_template.replace('{analisis}', analisis)
```

```python
llm = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash",
    google_api_key=os.getenv("GOOGLE_API_KEY"),
    max_retries=0
)

respuesta = llm.invoke(prompt)

html = respuesta.text

html
```

```python
dbutils.jobs.taskValues.set(key="html", value=html)
```
