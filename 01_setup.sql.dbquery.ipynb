```sql
-- Creo un banco de datos

CREATE DATABASE if not exists db_analytics;


-- Creo una tabla bronze para precio brent

CREATE TABLE if not exists db_analytics.bronze (
    fecha DATE,
    apertura DOUBLE,
    maximo DOUBLE,
    minimo DOUBLE,
    cierre DOUBLE,
    volumen DOUBLE
)
USING DELTA; 

-- Creo una tabla gold para precio brent

CREATE TABLE if not exists db_analytics.precio_semanal_brent(
    semana_inicio DATE,
    semana_fin DATE,
    apertura DOUBLE,
    maximo DOUBLE,
    minimo DOUBLE,
    var_pct DOUBLE,
    volumen_medio DOUBLE,
    cierre DOUBLE
)
USING DELTA;
```
